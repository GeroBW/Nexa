# Import Required Libraries
Import the necessary libraries, including pandas.

```python
import pandas as pd
```

# Load Data
Load the data into a pandas DataFrame.

```python
# Load the data into a pandas DataFrame
data = {
    'brands': [
        {
            'brandName': 'BrandA',
            'Categories': [
                {
                    'categoryName': 'Category1',
                    'sizes': [
                        {'size': 'S', 'regionalSizes': {'US': 'S', 'EU': 'M'}},
                        {'size': 'M', 'regionalSizes': {'US': 'M', 'EU': 'L'}}
                    ]
                },
                {
                    'categoryName': 'Category2',
                    'sizes': [
                        {'size': 'L', 'regionalSizes': {'US': 'L', 'EU': 'XL'}}
                    ]
                }
            ]
        },
        {
            'brandName': 'BrandB',
            'Categories': [
                {
                    'categoryName': 'Category3',
                    'sizes': [
                        {'size': 'S', 'regionalSizes': {'US': 'XS', 'EU': 'S'}}
                    ]
                }
            ]
        }
    ]
}

# Normalize the data
normalized_data = pd.json_normalize(data['brands'], 'Categories', ['brandName'])

# Explode sizes
exploded_sizes = normalized_data.explode('sizes')

# Normalize the sizes column
sizes_normalized = pd.json_normalize(exploded_sizes['sizes'])

# Combine the normalized data with the sizes data
final_data = exploded_sizes.drop(columns=['sizes']).join(sizes_normalized)

# Display the final DataFrame
final_data
```

# Reformat DataFrame
Reformat the DataFrame to ensure it meets the desired structure, with each item and size having its own row.

```python
# Reformat DataFrame

# Normalize the data
normalized_data = pd.json_normalize(data['brands'], 'Categories', ['brandName'])

# Explode sizes
exploded_sizes = normalized_data.explode('sizes')

# Normalize the sizes column
sizes_normalized = pd.json_normalize(exploded_sizes['sizes'])

# Combine the normalized data with the sizes data
final_data = exploded_sizes.drop(columns=['sizes']).join(sizes_normalized)

# Explode regionalSizes to separate rows
final_data = final_data.explode('regionalSizes')

# Normalize the regionalSizes column
regional_sizes_normalized = pd.json_normalize(final_data['regionalSizes'])

# Combine the final data with the regional sizes data
final_data = final_data.drop(columns=['regionalSizes']).join(regional_sizes_normalized)

# Display the final DataFrame
final_data
```
